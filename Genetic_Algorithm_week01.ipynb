```python
import random

# --------------------------------------------------
# 1. Define the Problem
# --------------------------------------------------

def fitness(x):
    """Fitness function to maximize."""
    return x ** 2


# --------------------------------------------------
# 2. Initialize Parameters
# --------------------------------------------------

POPULATION_SIZE = 10
CHROMOSOME_LENGTH = 5
MUTATION_RATE = 0.01
CROSSOVER_RATE = 0.8
GENERATIONS = 50


# --------------------------------------------------
# 3. Create Initial Population
# --------------------------------------------------

def create_population():
    """Generate a random initial population."""
    return [
        [random.randint(0, 1) for _ in range(CHROMOSOME_LENGTH)]
        for _ in range(POPULATION_SIZE)
    ]


# Convert binary chromosome to decimal value
def chromosome_to_decimal(chromosome):
    return int("".join(map(str, chromosome)), 2)


# --------------------------------------------------
# 4. Evaluate Fitness
# --------------------------------------------------

def evaluate_population(population):
    """Calculate fitness for every individual."""
    return [
        fitness(chromosome_to_decimal(individual))
        for individual in population
    ]


# --------------------------------------------------
# 5. Selection
# --------------------------------------------------

def selection(population, fitness_values):
    """
    Roulette-wheel selection.
    Individuals with higher fitness have a higher
    probability of being selected.
    """
    total_fitness = sum(fitness_values)

    if total_fitness == 0:
        return random.choice(population)

    probabilities = [
        value / total_fitness for value in fitness_values
    ]

    return random.choices(
        population,
        weights=probabilities,
        k=1
    )[0].copy()


# --------------------------------------------------
# 6. Crossover
# --------------------------------------------------

def crossover(parent1, parent2):
    """Single-point crossover."""
    if random.random() < CROSSOVER_RATE:
        point = random.randint(1, CHROMOSOME_LENGTH - 1)

        child1 = parent1[:point] + parent2[point:]
        child2 = parent2[:point] + parent1[point:]

        return child1, child2

    return parent1.copy(), parent2.copy()


# --------------------------------------------------
# 7. Mutation
# --------------------------------------------------

def mutate(chromosome):
    """Flip each bit with the given mutation probability."""
    for i in range(CHROMOSOME_LENGTH):
        if random.random() < MUTATION_RATE:
            chromosome[i] = 1 - chromosome[i]

    return chromosome


# --------------------------------------------------
# 8. Run the Genetic Algorithm
# --------------------------------------------------

def genetic_algorithm():
    population = create_population()

    best_solution = None
    best_fitness = float("-inf")

    for generation in range(GENERATIONS):

        # Evaluate fitness
        fitness_values = evaluate_population(population)

        # Track best solution
        generation_best_index = fitness_values.index(
            max(fitness_values)
        )

        generation_best = population[generation_best_index]
        generation_best_fitness = fitness_values[generation_best_index]

        if generation_best_fitness > best_fitness:
            best_fitness = generation_best_fitness
            best_solution = generation_best.copy()

        print(
            f"Generation {generation + 1:2d}: "
            f"Best x = {chromosome_to_decimal(generation_best)}, "
            f"Fitness = {generation_best_fitness}"
        )

        # Create next generation
        new_population = []

        while len(new_population) < POPULATION_SIZE:

            # Selection
            parent1 = selection(population, fitness_values)
            parent2 = selection(population, fitness_values)

            # Crossover
            child1, child2 = crossover(parent1, parent2)

            # Mutation
            child1 = mutate(child1)
            child2 = mutate(child2)

            new_population.append(child1)

            if len(new_population) < POPULATION_SIZE:
                new_population.append(child2)

        population = new_population

    return best_solution, best_fitness


# --------------------------------------------------
# 9. Output the Best Solution
# --------------------------------------------------

best_solution, best_fitness = genetic_algorithm()

best_x = chromosome_to_decimal(best_solution)

print("\n===== Final Result =====")
print("Best chromosome:", "".join(map(str, best_solution)))
print("Best solution x:", best_x)
print("Maximum fitness:", best_fitness)
```

```python
#Using Pyplot and seaborn to visualize
import random
import matplotlib.pyplot as plt
import seaborn as sns

# --------------------------------------------------
# 1. Define the Problem
# --------------------------------------------------

def fitness(x):
    return x ** 2


# --------------------------------------------------
# 2. Initialize Parameters
# --------------------------------------------------

POPULATION_SIZE = 10
CHROMOSOME_LENGTH = 5
MUTATION_RATE = 0.01
CROSSOVER_RATE = 0.8
GENERATIONS = 50


# --------------------------------------------------
# 3. Create Initial Population
# --------------------------------------------------

def create_population():
    return [
        [random.randint(0, 1) for _ in range(CHROMOSOME_LENGTH)]
        for _ in range(POPULATION_SIZE)
    ]


def chromosome_to_decimal(chromosome):
    return int("".join(map(str, chromosome)), 2)


# --------------------------------------------------
# 4. Evaluate Fitness
# --------------------------------------------------

def evaluate_population(population):
    return [
        fitness(chromosome_to_decimal(individual))
        for individual in population
    ]


# --------------------------------------------------
# 5. Selection
# --------------------------------------------------

def selection(population, fitness_values):

    total_fitness = sum(fitness_values)

    if total_fitness == 0:
        return random.choice(population)

    probabilities = [
        value / total_fitness for value in fitness_values
    ]

    return random.choices(
        population,
        weights=probabilities,
        k=1
    )[0].copy()


# --------------------------------------------------
# 6. Crossover
# --------------------------------------------------

def crossover(parent1, parent2):

    if random.random() < CROSSOVER_RATE:

        point = random.randint(1, CHROMOSOME_LENGTH - 1)

        child1 = parent1[:point] + parent2[point:]
        child2 = parent2[:point] + parent1[point:]

        return child1, child2

    return parent1.copy(), parent2.copy()


# --------------------------------------------------
# 7. Mutation
# --------------------------------------------------

def mutate(chromosome):

    for i in range(CHROMOSOME_LENGTH):

        if random.random() < MUTATION_RATE:
            chromosome[i] = 1 - chromosome[i]

    return chromosome


# --------------------------------------------------
# 8. Genetic Algorithm
# --------------------------------------------------

def genetic_algorithm():

    population = create_population()

    best_solution = None
    best_fitness = float("-inf")

    # Lists for visualization
    generation_numbers = []
    best_fitness_history = []
    best_x_history = []
    average_fitness_history = []

    for generation in range(GENERATIONS):

        # Evaluate fitness
        fitness_values = evaluate_population(population)

        # Find best individual
        generation_best_index = fitness_values.index(
            max(fitness_values)
        )

        generation_best = population[generation_best_index]

        generation_best_fitness = fitness_values[
            generation_best_index
        ]

        generation_best_x = chromosome_to_decimal(
            generation_best
        )

        # Update overall best
        if generation_best_fitness > best_fitness:

            best_fitness = generation_best_fitness

            best_solution = generation_best.copy()

        # Store data for visualization
        generation_numbers.append(generation + 1)
        best_fitness_history.append(generation_best_fitness)
        best_x_history.append(generation_best_x)
        average_fitness_history.append(
            sum(fitness_values) / len(fitness_values)
        )

        print(
            f"Generation {generation + 1:2d}: "
            f"Best x = {generation_best_x}, "
            f"Fitness = {generation_best_fitness}"
        )

        # Create next generation
        new_population = []

        while len(new_population) < POPULATION_SIZE:

            # Selection
            parent1 = selection(
                population,
                fitness_values
            )

            parent2 = selection(
                population,
                fitness_values
            )

            # Crossover
            child1, child2 = crossover(
                parent1,
                parent2
            )

            # Mutation
            child1 = mutate(child1)
            child2 = mutate(child2)

            new_population.append(child1)

            if len(new_population) < POPULATION_SIZE:
                new_population.append(child2)

        population = new_population

    return (
        best_solution,
        best_fitness,
        generation_numbers,
        best_fitness_history,
        best_x_history,
        average_fitness_history
    )


# --------------------------------------------------
# 9. Run Genetic Algorithm
# --------------------------------------------------

(
    best_solution,
    best_fitness,
    generations,
    best_fitness_history,
    best_x_history,
    average_fitness_history
) = genetic_algorithm()


# --------------------------------------------------
# 10. Final Result
# --------------------------------------------------

best_x = chromosome_to_decimal(best_solution)

print("\n===== Final Result =====")
print(
    "Best chromosome:",
    "".join(map(str, best_solution))
)
print("Best solution x:", best_x)
print("Maximum fitness:", best_fitness)


# --------------------------------------------------
# 11. Visualization
# --------------------------------------------------

sns.set_theme(style="whitegrid")


# Graph 1: Best Fitness
plt.figure(figsize=(10, 5))

sns.lineplot(
    x=generations,
    y=best_fitness_history,
    marker="o"
)

plt.title("Genetic Algorithm - Best Fitness per Generation")
plt.xlabel("Generation")
plt.ylabel("Best Fitness")
plt.xticks(range(1, GENERATIONS + 1, 5))
plt.show()


# Graph 2: Best X
plt.figure(figsize=(10, 5))

sns.lineplot(
    x=generations,
    y=best_x_history,
    marker="o"
)

plt.title("Genetic Algorithm - Best Solution (x)")
plt.xlabel("Generation")
plt.ylabel("Best x")
plt.xticks(range(1, GENERATIONS + 1, 5))
plt.show()


# Graph 3: Best vs Average Fitness
plt.figure(figsize=(10, 5))

sns.lineplot(
    x=generations,
    y=best_fitness_history,
    marker="o",
    label="Best Fitness"
)

sns.lineplot(
    x=generations,
    y=average_fitness_history,
    marker="s",
    label="Average Fitness"
)

plt.title("Genetic Algorithm - Fitness Progress")
plt.xlabel("Generation")
plt.ylabel("Fitness")
plt.xticks(range(1, GENERATIONS + 1, 5))
plt.legend()
plt.show()
```
